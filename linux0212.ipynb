문제2. 아래 설명 중 틀린것을 골라주세요

1. pwd : shell의 현재 작업 디렉토리에 대한 절대 경로를 표준 출력에 인쇄합니다.
2. `cd [<new location>]` : 셀의 현재 작업 디렉터리를 선택적인 <새 위치>로 변경합니다.
3. `ls [<location>]` 선택적 <location> 디렉터리의 내용을 나열합니다.
4. `ls [<location>]` <location>이 제공되지 않으면 셸의 이전 작업 디렉터리 내용을 인쇄합니다.

```python
#!/bin/bash

echo "문제 2: 틀린 설명을 확인합니다."

# 정답 저장 (틀린 설명)
wrong_statements=()

# 1번 검증: `pwd` 명령어는 현재 작업 디렉토리의 절대 경로를 출력함 (올바름)
if false; then
    wrong_statements+=("1")
fi

# 2번 검증: `cd [<new location>]` 명령어는 현재 작업 디렉토리를 변경함 (올바름)
if false; then
    wrong_statements+=("2")
fi

# 3번 검증: `ls [<location>]` 명령어는 선택적 <location> 디렉토리의 내용을 나열함 (올바름)
if false; then
    wrong_statements+=("3")
fi

# 4번 검증: `ls`는 기본적으로 현재 디렉토리 내용을 나열함, 이전 작업 디렉토리를 출력하지 않음 (틀림)
if true; then
    wrong_statements+=("4")
fi

# 결과 출력
if [ ${#wrong_statements[@]} -eq 0 ]; then
    echo "틀린 설명이 없습니다."
else
    echo "틀린 설명: ${wrong_statements[*]}"
fi
```

문제5. find와 locate 명령어에 대한 설명으로 옳지 않은 것을 골라주세요

1. locate 명렁어가 find 명령어보다 검색 속도가 빠르다
2. locate 명령어가 find 명령어보다 다양한 옵션을 가지고 있다.
3. locate 명렁어가 find 명령어보다 부정확한 데이터를 출력할 가능성이 있다.
4. Find 명령어는 단순히 파일을 찾는것 뿐만 아니라 찾은 파일에 대하여 특정 작업까지도 수행할 수 있다.

```python
#!/bin/bash

echo "문제 5: locate와 find 명령어에 대한 옳지 않은 설명을 확인합니다."

# 정답 저장 (틀린 설명)
wrong_statements=()

# 1번 검증: locate 명령어가 find 명령어보다 검색 속도가 빠름 (올바름)
if false; then
    wrong_statements+=("1")
fi

# 2번 검증: locate는 find보다 다양한 옵션을 제공하지 않음 (틀림)
if true; then
    wrong_statements+=("2")
fi

# 3번 검증: locate는 데이터베이스 기반으로 검색하므로 부정확한 데이터를 출력할 가능성 있음 (올바름)
if false; then
    wrong_statements+=("3")
fi

# 4번 검증: find는 파일 찾기뿐만 아니라 작업을 수행할 수 있음 (올바름)
if false; then
    wrong_statements+=("4")
fi

# 결과 출력
if [ ${#wrong_statements[@]} -eq 0 ]; then
    echo "틀린 설명이 없습니다."
else
    echo "틀린 설명: ${wrong_statements[*]}"
fi
```

문제7. 폴더명이 2023_01, 2023_02, 2023_03 ... 2024_12인 형태로 한번에 생성되도록 linux 명령어를 작성해주세요

```python
#!/bin/bash

echo "2023_01부터 2024_12까지 폴더를 한번에 생성합니다."

# seq 명령어와 for 루프를 사용하여 폴더 생성
for year in {2023..2024}; do
    for month in {01..12}; do
        mkdir "${year}_${month}"
    done
done
```
